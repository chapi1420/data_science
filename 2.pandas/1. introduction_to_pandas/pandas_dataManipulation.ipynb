## data manipulation using pandas

## missing values handling
we now have an dataframe with missing values included. This is how the real world operates and we will see the ways we could handle the missing values

```python
car_sales_missing  = pd.read_csv("data/initialManipulation.csv")
```

```python
car_sales_missing 
```

#### fillna 
this is used to fill the missing values with statistical estimate to continue using the data. For numerical data, we could use the mean or median, or maybe 
quadrant values too, for categorical values we could use modes to replace the missing values. 

```python
car_sales_missing =car_sales_missing.drop(columns= "Unnamed: 0", axis=1)
```

```python
car_sales_missing
```

```python
car_sales_missing["odometer (km)"].fillna(car_sales_missing['odometer (km)'].mean(), inplace=True)
```

```python
car_sales_missing
```

```python
#replace the categorical NaN with a mode value
car_sales_missing.colour.fillna(car_sales_missing['colour'].mode()[0], inplace=True)
```

key note on how **mode** works
mode works in a manner a bit different from the numerical center of tendency measures. the mode() function returns al lisst of values and
fillna doesnt know what to do with it, so it does nothing. we have to tell the fillna function to take the first index (0) in order to get the one with the highest value.

```python
car_sales_missing
```

### dropna function
unlike fillna function, dropna is used to drop entries that have a NaN value in them. This method is efficient if too many attributes of a tuple are missing and it is disadvantageous to have the tuple compared to the risk and misiformation it could bring.

```python
car_sales_missing =pd.read_csv("data/initialManipulation.csv")
```

```python
car_sales_missing.dropna(inplace=True)
```

```python
car_sales_missing.drop("Unnamed: 0", axis =1)
```

we can see using the indexes that rows(entries) with missing entries(NaN ) are removed from the dataframe. we can now save the cleaned data in another csv

```python
car_sales_missing.to_csv("data/car_sales_missing_handled.csv")
```
